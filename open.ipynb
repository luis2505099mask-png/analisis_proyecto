# Algoritmo de luhn 

El algoritmo de luhn es usado para verificar la validez de cosas como tarjetas o identificaciónes
Funciona selecionando de derecha a izquierda y duplicando su segundo digito, si el numero es divisible entre 10 es valido.

```python
card_num = input("Ingrese el número de su tarjeta: ")

def luhn_check(card_num):
    # Eliminar espacios en blanco y convertir a lista de dígitos
    digits = [int(x) for x in str(card_num).replace(" ", "")][::-1]
    
    # Aplicar el algoritmo de Luhn
    for i in range(1, len(digits), 2):
        digits[i] *= 2
        if digits[i] > 9:
            digits[i] -= 9
    
    # Calcular la suma total
    total_sum = sum(digits)
    
    # Verificar si el número es válido
    if total_sum % 10 == 0:
        print ("~" *22)
        print(f"El número de tarjeta {card_num} es válido.")
        print ("~" *22)
    else:
        print ("~" *22)
        print(f"El número de tarjeta {card_num} no es válido.")
        print ("~" *22)

luhn_check(card_num)
```

# Algoritmo con numpy

Busque un algoritmo y encontre uno que sirve para vectorizar una funcion con arreglos, pero lo modificare, pedire que el usuario de un numero y que de un numero mas grande que ese numero, el algoritmo tomara 5 numeros aleatorios y hara esto, si el numero es mayor a 10 lo multiplica por 2, si es 10 o menor que 10 lo multiplica por 3

```python
import numpy as np
```

```python
p = int(input("Ingrese un numero menor a 11"))
t = int(input("Ingrese un numero mayor a 10"))
a = np.random.randint(p,t)

def func(a):
    if a >10:
        return a * 2
    else:
        return a * 3

func_vect = np.vectorize(func)

b = func_vect(a)
print("=" *20)
print(f"Variables; p: {p} t: {t}")
print("=" *22)
print(f"Numero sellecionado: {a}")
print("=" *22)
print(f"Resultado: {b}")
print("=" *22)
```

# Prueba 2 de numpy 

El algoritmo de euclides calcula la distancia entre dos arreglos de n dimensiones y obtenemos una linea recta corta entre ambos arreglos

```python
def dis_euc(punto_alpha, punto_beta):
    alpha = np.array(punto_alpha)
    beta = np.array(punto_beta)
    return np.sqrt(np.sum((alpha - beta)**2))

entrada1=input("Escribe las coordenadas euclidianas del primer punto separadas por espacio")
entrada2=input("Escribe las coordenadas euclidianas del seundo punto separadas por espacio")

punto1 = [float(x) for x in entrada1.strip().split()]
punto2 = [float(x) for x in entrada2.strip().split()]

if len(punto1)!=len(punto2):
    print("Dimensiones no validas")
else:
    resultado = dis_euc(punto1, punto2)
    print("--"*22)
    print (f"Primer punto: {punto1} Segundo punto: {punto2}")
    print("--"*22)
    print(f"Distancia euclidiana; {resultado:.4f}")
    print("--"*22)
```
