## Generación y formateo de matrices aleatorias con Python puro

## Funcionamiento del código

### 1. Importación de la biblioteca `random`

Se utiliza el módulo `random` para generar números enteros aleatorios dentro de un rango específico.

### 2. Función `generar_matriz(N)`

Esta función crea una matriz cuadrada de tamaño `N × N`.

* Inicializa una lista vacía llamada `matriz`.
* Recorre `N` filas mediante un ciclo `for`.
* Para cada fila, genera `N` números aleatorios entre 1 y 9 utilizando `random.randint(1, 9)`.
* Agrega cada fila a la matriz principal.
* Finalmente devuelve la matriz completa.

Por ejemplo, para `N = 3`, la salida podría ser:

```text
[4, 2, 8]
[1, 7, 3]
[9, 5, 6]
```

### 3. Impresión de una matriz generada

Se genera una matriz de tamaño `5 × 5` y se imprime fila por fila para visualizar su contenido.

### 4. Función `imprimir_matriz(matriz)`

Esta función mejora la presentación de la matriz.

* Recorre cada fila de la matriz.
* Convierte cada elemento en texto con un ancho fijo de dos caracteres.
* Une los elementos mediante espacios para mantener una alineación uniforme.
* Imprime cada fila formateada.

Esto permite visualizar matrices grandes de manera más ordenada y legible.

### 5. Generación de matrices de distintos tamaños

El programa crea y muestra tres matrices cuadradas:

* Matriz de `5 × 5`
* Matriz de `10 × 10`
* Matriz de `20 × 20`

Entre cada una se imprime una línea separadora para facilitar la lectura de los resultados.

## Objetivo

Este ejercicio demuestra cómo:

* Utilizar ciclos anidados para construir estructuras bidimensionales.
* Generar datos aleatorios en Python.
* Trabajar con matrices representadas mediante listas de listas.
* Formatear la salida para mejorar la visualización de los datos.
* Escalar la generación de matrices a diferentes tamaños sin modificar la lógica principal del programa.

Este tipo de técnicas son fundamentales en programación, simulaciones, análisis de datos y algoritmos que requieren manipular estructuras matriciales.


````python
import random
````

````python
def generar_matriz(N):
    matriz= []
    for i in range (N):
        fila = []
        for j in range(N):
            fila.append(random.randint(1, 9))
        matriz.append(fila)
    return matriz

N = 5

matriz_aleatoria = generar_matriz(N)
for fila in matriz_aleatoria:
    print(fila)

def imprimir_matriz(matriz):
    for fila in matriz:
        fila_formateada = ' '.join(f'{elemento:2}' for elemento in fila)
        print(fila_formateada)
````

````python
print("Matriz de 5x5")

m_5 = generar_matriz(5)
imprimir_matriz(m_5)
print("=" *30)
print("Matriz de 10x10")
m_10 = generar_matriz(10)
imprimir_matriz(m_10)
print("=" *30)
print("Matriz de 20x20")
m_20 = generar_matriz(20)
imprimir_matriz(m_20)
````
