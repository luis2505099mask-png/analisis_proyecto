## Creación y análisis básico de un arreglo con NumPy

Este código muestra cómo convertir una lista de listas de Python en un arreglo de NumPy (`ndarray`) y cómo consultar algunas de sus propiedades fundamentales.

Primero, se define una estructura de datos con varias filas y columnas y se convierte a un arreglo utilizando `np.array()`, especificando que todos los elementos deben almacenarse como números de tipo `float`.

Posteriormente, se inspeccionan algunas características del arreglo:

* **`ndim`**: Indica el número de dimensiones del arreglo. En este caso, el resultado es `2`, ya que se trata de una matriz con filas y columnas.
* **`shape`**: Muestra la forma del arreglo en el formato `(filas, columnas)`. Para este ejemplo, la salida es `(3, 4)`.
* **`size`**: Devuelve la cantidad total de elementos contenidos en el arreglo. Al tener 3 filas y 4 columnas, el total es `12`.

Este tipo de operaciones son fundamentales en el análisis de datos y la computación científica, ya que permiten comprender la estructura y organización de los datos antes de realizar cálculos o transformaciones más avanzadas.

```python
import numpy as np
```

```python
lista = [[1, 2, 3, 4],[3, 4, 5, 6], [5, 6, 7, 8]]

array_1d = np.array(lista, dtype='float')


#Checamos el tipo de dato de array_id
type(array_1d)

#contenido del array
array_1d
```

```python
# ndim
print('Dimensiones: ', array_1d.ndim)
# shape
print("Shape: ", array_1d.shape)
# size
print("Size: ", array_1d.size)
```
